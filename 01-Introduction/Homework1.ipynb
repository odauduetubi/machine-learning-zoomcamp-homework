## Q6

1. Find the median value of the `horsepower` column in the dataset.
2. Next, calculate the most frequent value of the same `horsepower` column.
3. Use the `fillna` method to fill the missing values in the `horsepower` column with the most frequent value from the previous step.
4. Now, calculate the median value of `horsepower` once again.
   
What did you observe?

```python
df.horsepower.median()

print(f"Median horsepower of cars: {df.horsepower.median()}")
```

```python
df.horsepower.mode()[0]

print(f"Modal horsepower of cars: {df.horsepower.mode()[0]}")
```

```python
df['horsepower'].fillna(df['horsepower'].mode()[0]) #Filling the missing values in the horsepower column with the mode
```

```python
df['horsepower'].fillna(df['horsepower'].mode()[0]).median()

print(f"Median after filling missing values: {df['horsepower'].fillna(df['horsepower'].mode()[0]).median()}")
```

From the above, we see that the median value *reduced* after replacing the missing values by the mode.

## Q7
1. Select all the cars from Asia
2. Select only columns `vehicle_weight` and `model_year`
3. Select the first 7 values
4. Get the underlying NumPy array. Let's call it `X`.
5. Compute matrix-matrix multiplication between the transpose of `X` and `X`. To get the transpose, use `X.T`. Let's call the result `XTX`.
6. Invert `XTX`.
7. Create an array `y` with values `[1100, 1300, 800, 900, 1000, 1100, 1200]`.
8. Multiply the inverse of `XTX` with the transpose of `X`, and then multiply the result by `y`. Call the result `w`.
9. What's the sum of all the elements of the result?

```python
df[df['origin'] == 'Asia'] # A selection of all cars from Asia.
```

```python
df[df['origin'] == 'Asia'][['vehicle_weight', 'model_year']] #Selecting just the vehicle_wight and model_year column
```

```python
df[df['origin'] == 'Asia'][['vehicle_weight', 'model_year']].head(7) #Filtering the above to the first 7 entries
```

```python
X = df[df['origin'] == 'Asia'][['vehicle_weight', 'model_year']].head(7).values # Putting the above in a numpy array
X
```

```python
print(X.T)
```

```python
XTX = X.T.dot(X)

XTX
```

```python
XTX_inv = np.linalg.inv(XTX)
XTX_inv
```

```python
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])
y
```

```python
m = XTX_inv.dot(X.T)
m
```

```python
w = m.dot(y)

w
```

```python
print(w[0] + w[1])
```
