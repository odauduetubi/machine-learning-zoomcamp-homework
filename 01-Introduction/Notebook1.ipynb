## Linear Algebra

````python
u = np.array([2, 4, 5, 6])
u
````

````python
u.shape
````

````python
u.shape[0]
````

````python
v = np.array([1, 0, 0, 2])
v
````

````python
def vector_vector_multiplication(u, v):
    assert u.shape[0] == v.shape[0]

    n = u.shape[0]

    result = 0.0

    for i in range(n):
        result = result + u[i] * v[i]

    return result
````

````python
vector_vector_multiplication(u, v)
````

````python
u.dot(v)
````

### Matrix Vector multiplication

````python
P = np.array([[2, 4, 5, 6],
              [1, 2, 1, 2],
              [3, 1, 2, 1]])
P
````

````python
P.shape
````

````python
q = np.array([1, 0.5, 2, 1])
q
````

````python
def matrix_vector_multiplication(P, q):
    assert P.shape[1] == q.shape[0]

    num_rows = P.shape[0]

    result = np.zeros(num_rows)

    for i in range(num_rows):
        result[i] = vector_vector_multiplication(P[i], q)

    return result
    
````

````python
matrix_vector_multiplication(P, q)
````

````python
P.dot(q)
````

### Matrix-matrix multiplication

````python
Q = np.array([[1, 1, 2],
              [0, 0.5, 1],
              [0, 2, 1],
              [2, 1, 0]])
Q
````

````python
def matrix_matrix_multiplication(P, Q):
    assert P.shape[1] == Q.shape[0]

    num_rows = P.shape[0]
    num_cols = Q.shape[1]

    result = np.zeros((num_rows, num_cols))

    for i in range(num_cols):
        qi = Q[:, i]
        Pqi = matrix_vector_multiplication(P, qi)

        result[:, i] = Pqi


    return result
        
    
    
````

````python
matrix_matrix_multiplication(P, Q)
````

````python
P.dot(Q)
````
